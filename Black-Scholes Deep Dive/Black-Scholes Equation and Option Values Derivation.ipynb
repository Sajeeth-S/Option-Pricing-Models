<h1><center><ins>Deriving the <b>Black-Scholes Equation</b></ins></center></h1>

We will use V to represent the value of the option, and it will be a function of the value of the underlying asset S at current time t.

$V$ also is dependent on various other factors such as $\sigma$ representing asset price volatility, $K$ and $T$ representing the strike price and date of expiry respectively and $r$ representing the risk-free rate of interest.

So we can write $V$ as $V(S,t,\sigma,K,T,r)$

Let $\Pi$ denote the value of a portfolio of one long option position and a short position in
some quantity , $\Delta$, of the underlying:
$$\Pi = V(S,t) - \Delta S \tag{1}$$

The first term on the right is the option and the second term is the short asset position. Notice
the minus sign in front of the second term. The quantity  will for the moment be some
constant quantity of our choosing. We will assume that the underlying follows a lognormal
random walk:

$$dS = \mu Sdt + \sigma SdX$$

Let us consider how the value of the portfolio changes from time t to t + dt. The change in the portfolio value is due partly to the change in the option value and partly to the change in the underlying:
$$d\Pi = dV - \Delta dS \tag{2}$$

Notice that  has not changed during the time step; we have not anticipated the change in S.
From Ito's Lemma we have:

$$dV = \frac{\partial V}{\partial t}dt + \frac{\partial V}{\partial S}dS + \frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}V}{\partial S^{2}}dt \tag{3}$$

Therefore substituting (3) into (2),
$$d\Pi = \frac{\partial V}{\partial t}dt + \frac{\partial V}{\partial S}dS + \frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}V}{\partial S^{2}}dt - \Delta dS \tag{4}$$

The right-hand side of (4) contains two types of terms, the deterministic and the random. The deterministic terms are those with the $dt$, and the random terms are those with the $dS$. Pretending for the moment that we know V and its derivatives then we know everything about the right-hand side of (4) except for the value of $dS$. And this quantity we can never know in advance.
These random terms are the risk in our portfolio. Is there any way to reduce or even eliminate this risk? This can be done in theory (and almost in practice) by carefully choosing. The random terms in (4) are
$$\left(\frac{\partial V}{\partial S} - \Delta\right) dS$$
And thus to reduce randomness to 0 (delta hedging), let $$\Delta = \frac{\partial V}{\partial S} \tag{5}$$

By the principle of **No Arbitrage**, due to the riskless change in our portfolio over a given time, the change (denoted as $d\Pi$ must be equivalent to the growth we would get if we put the equivalent amount
of cash in a risk-free interest-bearing account. This implies that

$$d\Pi = r \Pi dt \tag{6}$$

Substituting (1),(4),(5) into (6), we get
$$\left(\frac{\partial V}{\partial t} + \frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}V}{\partial S^{2}}\right)dt = r \left(V - S\frac{\partial V}{\partial S}\right)dt$$
And dividing by $dt$ and rearranging,
$$ \Large \boxed{\frac{\partial V}{\partial t} + \frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}V}{\partial S^{2}} + rS\frac{\partial V}{\partial S} - rV = 0} \tag{7}$$

This is the derivation of the **Black-Scholes Equation** (7)

<h1><center><ins><b>Deriving the Black-Scholes Values for European and Binary Call/Put Options</b></ins></center></h1>

We will start by making substitutions to transform the original Black-Scholes Equation into a variant of a heat diffusion PDE

The first step in the manipulation is to change from present value to future value terms. Recalling that the payoff is received at time $T$ but that we are valuing the option at time $t$ this
suggests that we write

$$V(S,t) = e^{-r(T-t)}U(S,t) \tag{8}$$
And so,

$$\frac{\partial V}{\partial t}=e^{-r\,(T-t)}\left(rU(S,t)+\frac{\partial U}{\partial t}\right) , \frac{\partial V}{\partial S}=e^{-r\,(T-t)}\frac{\partial U}{\partial S} , \frac{\partial^{2}V}{\partial S^{2}}=e^{-r\,(T-t)}\frac{\partial^{2}U}{\partial S^{2}} $$
Substituting back into (8) we get, 

$$ e^{-r\,(T-t)}\left(rU(S,t)+\frac{\partial U}{\partial t}\right) + \frac{1}{2}\sigma^{2}S^{2}e^{-r\,(T-t)}\frac{\partial^{2}U}{\partial S^{2}} + rSe^{-r\,(T-t)}\frac{\partial U}{\partial S} - re^{-r\,(T-t)}U(S,t) = 0$$
Cancelling terms and dividing by $e^{-r\,(T-t)}$ gives, 

$$ \frac{\partial U}{\partial t} + \frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}U}{\partial S^{2}} + rS\frac{\partial U}{\partial S} = 0 \tag{9}$$

Now we will use our second substitution, $$\tau = T-t$$
Consequently, $t=T-\tau$, $\frac{\partial t}{\partial \tau} = -1$ and so, $\frac{\partial U}{\partial \tau} = \frac{\partial U}{\partial t}\frac{\partial t}{\partial \tau} = -\frac{\partial U}{\partial t}$.\
Thus substituting this into (9), $$\frac{\partial U}{\partial \tau} = \frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}U}{\partial S^{2}} + rS\frac{\partial U}{\partial S} \tag{10}$$

For our third substitution, $$ \xi = log{S} \iff e^{\xi} = S $$
Consequently, $\frac{\partial U}{\partial S} = \frac{\partial U}{\partial \xi}\frac{\partial \xi}{\partial S} = \frac{1}{S}\frac{\partial U}{\partial \xi} = e^{-\xi}\frac{\partial U}{\partial \xi}$
<br>
And, $\frac{\partial^{2}U}{\partial S^{2}} = \frac{\partial}{\partial S}\left(\frac{1}{S}\frac{\partial U}{\partial \xi}\right)=-\frac{1}{S^{2}}\frac{\partial U}{\partial \xi}+\frac{1}{S}\frac{\partial^{2} U}{\partial \xi \partial S}=-\frac{1}{S^{2}}\frac{\partial U}{\partial \xi}+\frac{1}{S}\frac{\partial^{2} U}{\partial \xi^{2}}\frac{\partial \xi}{\partial S}=\frac{1}{S^{2}}\frac{\partial^{2} U}{\partial \xi^{2}}-\frac{1}{S^{2}}\frac{\partial U}{\partial \xi}=e^{-2\xi}\frac{\partial^{2} U}{\partial \xi^{2}}-e^{-2\xi}\frac{\partial U}{\partial \xi}$
<br>
Substituting these into (10) we get,
$$\frac{\partial U}{\partial \tau}=\frac{1}{2}\sigma^{2}\left(\frac{\partial^{2} U}{\partial \xi^{2}}-\frac{\partial U}{\partial \xi}\right)+r\frac{\partial U}{\partial \xi}$$ $$\frac{\partial U}{\partial \tau}=\frac{1}{2}\sigma^{2}\frac{\partial^{2} U}{\partial \xi^{2}}+\left(r-\frac{1}{2}\sigma^{2}\right)\frac{\partial U}{\partial \xi}$$

Lastly, let
$$x=\xi + \left(r-\frac{1}{2}\sigma^{2}\right)\tau$$
After changing variables and letting $U=W(x,\tau)$, we get our intended PDE

$$\frac{\partial W}{\partial \tau}=\frac{1}{2}\sigma^{2}\frac{\partial^{2} W}{\partial x^{2}} \tag{11}$$

To directly link all our substitutions together, we can express it as follows
$$V(S,t)=e^{-r(T-t)}U(S,t)=e^{-r\tau}U(S,T-\tau)=e^{-r\tau}U(e^{\xi},T-\tau)=e^{-r\tau}U(e^{x-(r-\frac{1}{2}\sigma^{2})\tau},T-\tau)=e^{-r\tau}W(x,\tau) \tag{12}$$

We will attempt to solve this PDE by firstly finding the **fundamental solution** and then the **general solution**

Let us try to find a special solution of the form
$$W(x,\tau)=\tau^{\alpha}f\left(\frac{x-x'}{\tau^{\beta}}\right)=W_f(x,\tau;x') \tag{13}$$
Note that f is only dependent on **one** variable, $\frac{x-x'}{\tau^{\beta}}$, meaning that we will be left with a simpler ODE and we must find values of $\alpha, \beta$ that work.\
Substituting (13) into (11),
$$ \alpha \tau^{\alpha -1}f + \tau^{\alpha}\frac{\partial f}{\partial \tau}=\frac{1}{2}\sigma^{2}\tau^{\alpha}\frac{\partial ^{2}f}{\partial x^{2}} \tag{14}$$
In order to make f take one variable, let
$$\eta = \frac{x-x'}{\tau^{\beta}}$$
Transforming, note that $\frac{\partial f}{\partial \tau}=\frac{df}{d\eta}\frac{d\eta}{d\tau}=-\beta\frac{x-x'}{\tau^{\beta+1}}\frac{d\eta}{d\tau}=-\beta\frac{\eta}{\tau}\frac{d\eta}{d\tau}$\
And, $\frac{\partial f}{\partial x}=\frac{df}{d\eta}\frac{d\eta}{dx}=\frac{1}{\tau^{\beta}}\frac{df}{d\eta}$ , $\frac{\partial^{2} f}{\partial x^{2}}=\frac{\partial}{\partial x}\left(\frac{\partial f}{\partial x}\right)=\frac{1}{\tau^{\beta}}\frac{d^{2}f}{d\eta^{2}}\frac{d\eta}{dx}=\frac{1}{\tau^{2\beta}}\frac{d^{2}f}{d\eta^{2}}$\
Substituting into (14),
$$\tau^{\alpha-1}\left(\alpha f - \beta\eta\,\frac{df}{d\eta}\right)=\frac{1}{2}\sigma^{2}\tau^{\alpha-2\beta}\frac{d^{2}f}{d\eta^{2}} \tag{15}$$\
Looking at both sides of the equation and the terms involving $\tau$, we can see that a solution holds only if
$$\alpha-1=\alpha-2\beta \implies \beta=\frac{1}{2}$$\
Now we want to find a value of $\alpha$ such that the integral of our fundamental solution over all $\xi$ is independent of $\tau$. Using the fact that $x-x'=\tau^{\beta}\eta \implies dx=\tau^{\beta}d\eta$,
$$\int_{-\infty}^{\infty}\tau^{\alpha}f\left(\frac{x-x'}{\tau^{\beta}}\right)\,dx=\int_{-\infty}^{\infty}\tau^{\alpha+\beta}f(\eta)\,d\eta=\tau^{\alpha+\beta}\int_{-\infty}^{\infty}f(\eta)\,d\eta$$
In order to be independent of $\tau$, $$\alpha+\beta=0 \iff \alpha=-\beta \implies \alpha=-\frac{1}{2}$$\
With our values of $\alpha,\beta$ and substituting into (15)
$$\tau^{-\frac{3}{2}}\left(-\frac{1}{2}f-\frac{1}{2}\eta\frac{df}{d\eta}\right)=\frac{1}{2}\sigma^{2}\tau^{-\frac{3}{2}}\frac{d^{2}f}{d\eta^{2}}$$
$$\implies -\,f-\,\eta\frac{df}{d\eta}=\sigma^{2}\frac{d^{2}f}{d\eta^{2}}$$\
By Product Rule,$$\implies \,\sigma^{2}\frac{d^{2}f}{d\eta^{2}}+\frac{d(\eta f)}{d\eta}=0$$\
By integrating once,$$\implies \,\sigma^{2}\frac{df}{d\eta}+\eta f=c_1=0$$ Where $c_1$ is our constant which we choose to be 0 for our fundamental solution.\
Solving this simple ODE via an integrating factor$=e^{\int\frac{\eta}{\sigma^{2}}\,d\eta}=e^{\frac{\eta^{2}}{2\sigma^{2}}}$
$$\implies \,f(\eta)=c_2e^{-\eta^{2}\,/(2\sigma^{2})}$$
Now we will choose $c_2$ such that the integral of $f$ from $-\infty$ to $\infty$ is 1. Recalling that $\int_{-\infty}^{\infty}e^{-x^{2}}\,dx=\sqrt{\pi}$
Thus for $x=\frac{\eta}{\sqrt{2}\sigma}$, $$\int_{-\infty}^{\infty}e^{-\left(\frac{\eta}{\sqrt{2}\sigma}\right)^{2}}\,\frac{1}{\sqrt{2}\sigma}\,d\eta=\sqrt{\pi}$$
$$\implies c_2=\frac{1}{\sqrt{2\pi}\,\sigma}$$
$$\implies f(\eta)=\frac{1}{\sqrt{2\pi}\,\sigma}e^{-\eta^{2}\,/(2\sigma^{2})}$$
And,
$$W_f(x,\tau;x')=\tau^{\alpha}f\left(\frac{x-x'}{\tau^{\beta}}\right)=\tau^{-\frac{1}{2}}f(\eta)=\frac{1}{\sqrt{2\pi\,\tau}\,\sigma}e^{-((x-x')^{2}/2\sigma^{2}\tau)}$$
This is the **fundamental solution** we have been aiming for.\
Now we will graph it for various values of $\tau$ to see what we are working with.

```python
#Code to plot W_f as a function of (x-x') for various values of tau
import matplotlib.pyplot as plt
import numpy as np
def graph(tau,sigma,x1):
    x = np.linspace(-2,4,1000)
    w = (1/((np.sqrt(2*np.pi*tau))*sigma))*np.exp(-(((x-x1)**2)/(2*(sigma**2)*tau)))
    plt.plot(x, w, label=f'tau = {tau}')

graph(1,1,1)
graph(0.1,1,1)
graph(0.01,1,1)
graph(0.001,1,1)

plt.title('Graph of W_f')
plt.xlabel('x')
plt.ylabel('W_f')
plt.ylim(0,13)
plt.legend(loc='upper left')
plt.grid()
plt.show()
```

This graph shows that our function rapidly increases at the value $x=x'$ but decays at either side of this value. So as $\tau \rightarrow 0$, $W_f \rightarrow \infty$ when $x=x'$ and $W_f \rightarrow 0$ for all other values of x.\
Now remember how we previously chose our fundamental solution such that the integral of it from $-\infty$ to $\infty$ is 1. This means that for all differing values of $\tau$, the area under the curve is fixed at 1.\
This results in a **Dirac Delta Function** $\delta(x-x')$ as $\tau \rightarrow 0$. And via the **Sifting Property** for any point below x to any point above x,
$$\int\delta(x-x')g(x')\,dx'=g(x)$$
If we consider the limit as $\tau \rightarrow 0$, our function $W_f$ becomes this Delta Function at $x=x'$. So,
$$\lim_{\tau \to 0}\frac{1}{\sqrt{2\pi\,\tau}\,\sigma}\int_{-\infty}^{\infty}e^{-\frac{(x-x')^{2}}{2\sigma^{2}\,\tau}}g(x')\,dx'=g(x) \tag{16}$$


Now we will consider the Payoff and include that into our functions.\
The payoff is the value of the option at time $t=T$ so we can say that 
$$V(S,T)=\text{Payoff}(S)$$
Considering our transformations in (12), this is equivalent to
$$W(x,0)=\text{Payoff}(e^{x}) \tag{17}$$
Looking at (x), I will claim that the solution of this for $\tau > 0$ is
$$W(x,\tau)=\int_{-\infty}^{\infty} W_f\,(x,\tau;x')\text{Payoff}(e^{x'})\,dx'$$
We can check if this works via verifying that it is a valid solution to (11) and that it fits our final condition (17).
<br>
Firstly let us check that it is a valid solution to the PDE (11).

$$W(x,\tau)=\int_{-\infty}^{\infty} \frac{1}{\sqrt{2\pi\,\tau}\,\sigma}e^{-((x-x')^{2}/2\sigma^{2}\tau)}\text{Payoff}(e^{x'})\,dx'$$

$$\frac{\partial W}{\partial \tau}=\int_{-\infty}^{\infty} \frac{\partial}{\partial \tau}\left(\frac{1}{\sqrt{2\pi\,\tau}\,\sigma}e^{-((x-x')^{2}/2\sigma^{2}\tau)}\text{Payoff}(e^{x'})\right)\,dx'$$

$$= \int_{-\infty}^{\infty}-\frac{1}{2\sqrt{2\pi}\sigma\tau^{3/2}}e^{-((x-x')^{2}/2\sigma^{2}\tau)} - \frac{(x-x')^{2}}{2\sqrt{2\pi}\sigma^{3}\tau^{5/2}}e^{-((x-x')^{2}/2\sigma^{2}\tau)}\text{Payoff}(e^{x'})\,dx'$$

$$ = -\frac{1}{2}\frac{1}{\sqrt{2\pi\,\tau}\,\sigma}\int_{-\infty}^{\infty}\left(\frac{1}{\tau}+\frac{(x-x')^{2}}{\sigma^{2}\tau^{2}}\right)e^{-((x-x')^{2}/2\sigma^{2}\tau)}\text{Payoff}(e^{x'})\,dx'$$

$$\frac{\partial W}{\partial x}=\frac{1}{\sqrt{2\pi\,\tau}\,\sigma}\int_{-\infty}^{\infty}-\frac{x-x'}{\sigma^{2}\tau}e^{-((x-x')^{2}/2\sigma^{2}\tau)}\text{Payoff}(e^{x'})\,dx'$$

$$\frac{\partial^{2} W}{\partial x^{2}}=\frac{1}{\sqrt{2\pi\,\tau}\,\sigma}\int_{\infty}^{\infty}-\frac{1}{\sigma^{2}\tau}e^{-((x-x')^{2}/2\sigma^{2}\tau)}-\frac{(x-x')^{2}}{\sigma^{4}\tau^{2}}e^{-((x-x')^{2}/2\sigma^{2}\tau)}\text{Payoff}(e^{x'})\,dx'$$

$$=-\frac{1}{\sigma^{2}}\frac{1}{\sqrt{2\pi\,\tau}\,\sigma}\int_{-\infty}^{\infty}\left(\frac{1}{\tau}+\frac{(x-x')^{2}}{\sigma^{2}\tau^{2}}\right)e^{-((x-x')^{2}/2\sigma^{2}\tau)}\text{Payoff}(e^{x'})\,dx'$$

$$=\frac{2}{\sigma^{2}}\,\frac{\partial W}{\partial \tau}$$
$$\implies \frac{\partial W}{\partial \tau} = \frac{1}{2}\,\sigma^{2}\,\frac{\partial^{2} W}{\partial x^{2}}$$
Thereby, satisfying (11).
<br>
Now to check that the final condition is satisfied in (17),
$$W(x,0)= \lim_{\tau \to 0}\frac{1}{\sqrt{2\pi\,\tau}\,\sigma}\int_{-\infty}^{\infty}e^{-\frac{(x-x')^{2}}{2\sigma^{2}\,\tau}}\text{Payoff}(e^{x'})\,dx'$$
<br>
By the property of the Dirac Delta Function stated in (16)
$$\implies W(x,0)=\text{Payoff}(e^{x})$$
So this satisfies (17)
<br>
Therefore $W(x,\tau)$ is clearly a valid solution.
<br>
In order to get the solution back into our original variables from (12), recall that $x=\xi+\left(r-\frac{1}{2}\sigma^{2}\right)\tau$ , $\xi=\log{S}$ , $\tau=T-t$ and $V(S,t)=e^{-r(T-t)}U(S,t)$.

So,
<br>
$$\Large \boxed{V(S,t) = \frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}} \int_{-\infty}^{\infty} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}\text{Payoff}(e^{x'})\,dx'} \tag{18}$$\
Thus we have our exact solution for $V(S,t)$ that solves the **Black-Scholes Equation**.

<h1><center> <ins><b>Formula for a European Call Option Value</b></ins></center></h1>

Having derived an exact solution for the Black-Scholes Equation, we can move onto deriving the value of a European Call Option specifically. It is clear that a call option has the payoff function, $\text{Payoff}(S)=\text{max}\left(S-K,0\right)$, where $K$ is the **Strike/Exercise Price**. Knowing $\text{Payoff}(e^{x'})=\text{max}\left(e^{x'}-K,0\right)$ , letting $V_{EC}(S,t)$ be the Call Option Value of the underlying asset $S$ at time $t$ and using (18),
$$V_{EC}(S,t)=\frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}} \int_{-\infty}^{\infty} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}\text{max}\left(e^{x'}-K,0\right)\,dx'$$
Taking a look at the Payoff function, for $e^{x'}<K \iff x'<\log{K}$, we get 0 inside our integral since $e^{x'}-K<0$. Thus,
$$V_{EC}(S,t)=\frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}}\left( \int_{-\infty}^{\log{K}} 0\,dx' + \int_{log{K}}^{\infty} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}\left(e^{x'}-K\right)\,dx' \right)$$
$$=\frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}}\int_{log{K}}^{\infty} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}\left(e^{x'}-K\right)\,dx'$$
$$=\frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}} \left[ \int_{log{K}}^{\infty} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}e^{x'}\,dx' - K \int_{log{K}}^{\infty} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}\,dx'\right] = \frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}}\left(A-KB\right) = C-D \tag{19}$$
Where we have split the 2 terms into $C,D$ and involving $A,B$ to work on them separately.\
Looking at our 2 integrals, it may be obvious to spot the glaring similarity between these and the cumulative distribution function for the standardised Normal Distribution defined as 
$$N(x) = \frac{1}{\sqrt{2\pi}}\int_{-\infty}^{x} e^{-\frac{1}{2}\phi^{2}} d\phi \tag{20}$$
So let us attempt to get our integrals into this form.

$$A= \int_{log{K}}^{\infty} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}+x'}\,dx'$$
$$= \int_{log{K}}^{\infty} e^{-\frac{1}{2}\left\{ \frac{ \left(x'\right)^{2}-2x'\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)+\sigma^{2}\left(T-t\right)\right)+\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)\right)^{2}}{\sigma^{2}\,\left(T-t\right)} \right\}}\,dx'$$
$$= \int_{log{K}}^{\infty} e^{-\frac{1}{2}\left\{ \frac{ \left[x' - \left(\log{S}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)\right)\right]^{2}+\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)\right)^{2} - \left(\log{S}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)\right)^{2}}{\sigma^{2}\,\left(T-t\right)} \right\}}\,dx'$$
$$= \int_{log{K}}^{\infty} e^{-\frac{1}{2}\left\{ \frac{ \left[x' - \left(\log{S}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)\right)\right]^{2}+\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)\right)^{2} - \left(\log{S}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)\right)^{2}}{\sigma^{2}\,\left(T-t\right)} \right\}}\,dx'$$
$$= \int_{log{K}}^{\infty} e^{-\frac{1}{2}\left\{ \frac{ \left[x' - \left(\log{S}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)\right)\right]^{2} + \left(\log{S}\right)^{2}-\left(\log{S}\right)^{2} + \,\left(T-t\right)\left[2\log{S}\left(r-\frac{1}{2}\sigma^{2}\right)-2\log{S}\left(r+\frac{1}{2}\sigma^{2}\right)\right] + \,\left(T-t\right)^{2}\left[\left(r-\frac{1}{2}\sigma^{2}\right)^{2}-\left(r+\frac{1}{2}\sigma^{2}\right)^{2}\right]}{\sigma^{2}\,\left(T-t\right)} \right\}}\,dx'$$
$$= \int_{log{K}}^{\infty} e^{-\frac{1}{2}\left\{ \frac{ \left[x' - \left(\log{S}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)\right)\right]^{2} + \,\left(T-t\right)\left(-2\sigma^{2}\log{S}\right) + \,\left(T-t\right)^{2}\left(-2r\sigma^{2}\right)}{\sigma^{2}\,\left(T-t\right)} \right\}}\,dx'$$
$$= \int_{log{K}}^{\infty} e^{ \left\{ \frac{-\frac{1}{2}\left( x' - \left(\log{S}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)\right) \right)^{2} }{\sigma^{2}\,\left(T-t\right)} + \frac{ \left(T-t\right)\left(\sigma^{2}\log{S}\right) + \,\left(T-t\right)^{2}\left(r\sigma^{2}\right)}{\sigma^{2}\,\left(T-t\right)}\right\} }\,dx'$$
$$= \int_{log{K}}^{\infty} e^{ \frac{-\frac{1}{2}\left( x' - \left(\log{S}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)\right) \right)^{2} }{\sigma^{2}\,\left(T-t\right)}} \times e^{ \log{S}+r\left(T-t\right) } $$
$$= Se^{r\left(T-t\right)} \int_{log{K}}^{\infty} e^{ \frac{-\frac{1}{2}\left( x' - \left(\log{S}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)\right) \right)^{2} }{\sigma^{2}\,\left(T-t\right)}} \,dx'$$
We need to get this into the form of (20) so we require a substitution of variables and also in a way to change our bounds of integration. Firstly let us change our integrand using the fact that $(a-b)^{2} \equiv (b-a)^{2}$,
$$A= Se^{r\left(T-t\right)} \int_{log{K}}^{\infty} e^{ -\frac{1}{2}\left( \frac{\left( -x' + \log{S}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}\right)^{2} } \,dx'$$
Now let us use the substitution 
$$\phi= \frac{ -x' + \log{S}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{\sigma\,\sqrt{T-t}}$$
$$\implies d\phi=-\frac{dx'}{\sigma\,\sqrt{T-t}}$$
And as $x' \rightarrow \infty$, $\phi \rightarrow -\infty$. And for $x'=\log{K}$, $\phi = \frac{\left( -\log{K} + \log{S}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}=\frac{\left(\log{S/K}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}=d_1$.\
So,
<br>
$$A= Se^{r\left(T-t\right)} \int_{d_1}^{-\infty} -e^{-\frac{1}{2}\phi^{2}}\sigma\,\sqrt{T-t}d\phi$$
By using the fact that $\int_{a}^{b} -f(x)\,dx = \int_{b}^{a} f(x)\,dx$,
$$A = Se^{r\left(T-t\right)}\sigma\,\sqrt{T-t} \int_{-\infty}^{d_1} e^{-\frac{1}{2}\phi^{2}}d\phi$$
Using (19),
$$C= \frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}}A$$
$$= \frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}} Se^{r\left(T-t\right)}\sigma\,\sqrt{T-t} \int_{-\infty}^{d_1}e^{-\frac{1}{2}\phi^{2}}d\phi$$
$$= \frac{S}{\sqrt{2\pi}}\int_{-\infty}^{d_1}e^{-\frac{1}{2}\phi^{2}}d\phi$$
Using (20),
$$C = S N(d_1)$$

Now let us consider the second integral $B$ from (19).
$$B= \int_{log{K}}^{\infty} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}\,dx'$$
$$=\int_{log{K}}^{\infty} e^{-\frac{1}{2}\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{\sigma^{2}\,\left(T-t\right)}}\,dx' $$
$$=\int_{log{K}}^{\infty} e^{-\frac{1}{2}\left(\frac{-x'+\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{\sigma\,\sqrt{T-t}}\right)^{2}}\,dx' $$
Now let us use the substitution 
$$\phi= \frac{-x'+\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{\sigma\,\sqrt{T-t}}$$
$$\implies d\phi=-\frac{dx'}{\sigma\,\sqrt{T-t}}$$
And as $x' \rightarrow \infty$, $\phi \rightarrow -\infty$. And for $x'=\log{K}$, $\phi = \frac{\left( -\log{K} + \log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}=\frac{\left(\log{S/K}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}=d_2$.\
So,
<br>
$$B= \int_{d_2}^{-\infty} -e^{-\frac{1}{2}\phi^{2}}\sigma\,\sqrt{T-t}d\phi$$
By using the fact that $\int_{a}^{b} -f(x)\,dx = \int_{b}^{a} f(x)\,dx$,
$$B = \sigma\,\sqrt{T-t} \int_{-\infty}^{d_2} e^{-\frac{1}{2}\phi^{2}}d\phi$$
Using (19),
$$D= \frac{Ke^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}}B$$
$$= \frac{Ke^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}} \sigma\,\sqrt{T-t} \int_{-\infty}^{d_2}e^{-\frac{1}{2}\phi^{2}}d\phi$$
$$= \frac{Ke^{-r\left(T-t\right)}}{\sqrt{2\pi}}\int_{-\infty}^{d_2}e^{-\frac{1}{2}\phi^{2}}d\phi$$
Using (20),
$$= Ke^{-r\left(T-t\right)} N(d_2)$$

Putting this all together from (19), we get that 
$$V_{EC}(S,t)= C-D = S N(d_1) - Ke^{-r\left(T-t\right)} N(d_2)$$

$$\Large \implies \boxed{V_{EC}(S,t)=S N(d_1) - Ke^{-r\left(T-t\right)} N(d_2)}$$
where
$$d_1 = \frac{\log{S/K}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{\sigma\,\sqrt{T-t}} \tag{21}$$
and 
$$d_2 = \frac{\log{S/K}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{\sigma\,\sqrt{T-t}} \tag{22}$$
<br>
Also, note that $$d_2= \frac{\log{S/K}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-\sigma^{2}\left(T-t\right)}{\sigma\,\sqrt{T-t}}=d_1-\frac{\sigma^{2}\left(T-t\right)}{\sigma\,\sqrt{T-t}}$$
<br>
$$\implies d_2=d_1 - \sigma\sqrt{T-t}$$

<h1><center> <ins><b>Formula for a European Put Option Value</b></ins></center></h1>

We can use the same method we used for a European Call except this time, the Payoff Function is $\text{Payoff}(e^{x'})=\text{max}\left(K-e^{x'},0\right)$. Letting $V_{EP}(S,t)$ be the Put Option Value of the underlying asset $S$ at time $t$ and using (18),
$$V_{EP}(S,t)=\frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}} \int_{-\infty}^{\infty} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}\text{max}\left(K-e^{x'},0\right)\,dx'$$
Taking a look at the Payoff function, for $e^{x'}>K \iff x'>\log{K}$, we get 0 inside our integral since $K-e^{x'}<0$. Thus,
$$V_{EP}(S,t)=\frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}}\left( \int_{-\infty}^{\log{K}} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}\text{max}\left(K-e^{x'},0\right)\,dx'+\int_{\log{K}}^{\infty} 0\,dx' \right)$$
$$=\frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}} \int_{-\infty}^{\log{K}} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}\left(K-e^{x'}\right)\,dx'$$
$$=\frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}} \left[ - \int_{-\infty}^{\log{K}} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}} e^{x'} + K\int_{-\infty}^{\log{K}} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}\,dx'\right]$$
$$=\frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}}\left(-E+KF\right) = -G+H \tag{23}$$

Let us look at E separately, note the similarity between this integral and the integral $A$ in (19), where only the bounds of integration have changed. So from (23) we can say that 
$$E= Se^{r\left(T-t\right)} \int_{-\infty}^{\log{K}} e^{ -\frac{1}{2}\left( \frac{\left( -x' + \log{S}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}\right)^{2} } \,dx'$$
We cannot currently use the substitution we used last time since our bounds of integration would not be in the form we require so we will rearrange the numerator of the fraction being squared using the fact that $x^{2}=(-x)^{2}$. So,
$$E= Se^{r\left(T-t\right)} \int_{-\infty}^{\log{K}} e^{ -\frac{1}{2}\left( \frac{\left( x' - \log{S}-\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}\right)^{2} } \,dx'$$
Now let us use the substitution,
$$\phi = \frac{ x' - \log{S}-\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{\sigma\,\sqrt{T-t}}$$
$$\implies d\phi=\frac{dx'}{\sigma\,\sqrt{T-t}}$$
And as $x' \rightarrow -\infty$, $\phi \rightarrow -\infty$. And for $x'=\log{K}$, $\phi = \frac{\left( \log{K} - \log{S}-\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}=-\frac{\left(\log{S/K}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}=-d_1$. Where we are using $d_1$ from (21). So,
<br>
$$E= Se^{r\left(T-t\right)} \int_{-\infty}^{-d_1} e^{-\frac{1}{2}\phi^{2}}\sigma\,\sqrt{T-t}d\phi$$
$$E = Se^{r\left(T-t\right)}\sigma\,\sqrt{T-t} \int_{-\infty}^{-d_1} e^{-\frac{1}{2}\phi^{2}}d\phi$$
Using (23),
$$G= \frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}}E$$
$$= \frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}} Se^{r\left(T-t\right)}\sigma\,\sqrt{T-t} \int_{-\infty}^{-d_1}e^{-\frac{1}{2}\phi^{2}}d\phi$$
$$= \frac{S}{\sqrt{2\pi}}\int_{-\infty}^{-d_1}e^{-\frac{1}{2}\phi^{2}}d\phi$$
Using (20),
$$= S N(-d_1)$$

Let us look at F now, note the similarity between this integral and the integral $B$ in (19), where only the bounds of integration have changed. So from (23) we can say that 
$$F=\int_{-\infty}^{\log{K}} e^{-\frac{1}{2}\left(\frac{-x'+\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{\sigma\,\sqrt{T-t}}\right)^{2}}\,dx' $$
We cannot currently use the substitution we used last time since our bounds of integration would not be in the form we require so we will rearrange the numerator of the fraction being squared using the fact that $x^{2}=(-x)^{2}$. So,
$$F= \int_{-\infty}^{\log{K}} e^{ -\frac{1}{2}\left( \frac{\left( x' - \log{S}-\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}\right)^{2} } \,dx'$$
Now let us use the substitution,
$$\phi = \frac{ x' - \log{S}-\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{\sigma\,\sqrt{T-t}}$$
$$\implies d\phi=\frac{dx'}{\sigma\,\sqrt{T-t}}$$
And as $x' \rightarrow -\infty$, $\phi \rightarrow -\infty$. And for $x'=\log{K}$, $\phi = \frac{\left( \log{K} - \log{S}-\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}=-\frac{\left(\log{S/K}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}=-d_2$. Where we are using $d_2$ from (21).\
So,
<br>
$$F= \int_{-\infty}^{-d_2} e^{-\frac{1}{2}\phi^{2}}\sigma\,\sqrt{T-t}d\phi$$
$$= \sigma\,\sqrt{T-t} \int_{-\infty}^{-d_2} e^{-\frac{1}{2}\phi^{2}}d\phi$$
Using (23),
$$H= \frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}}K\,F$$
$$= \frac{Ke^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}} \sigma\,\sqrt{T-t} \int_{-\infty}^{-d_2}e^{-\frac{1}{2}\phi^{2}}d\phi$$
$$= \frac{Ke^{-r\left(T-t\right)}}{\sqrt{2\pi}}\int_{-\infty}^{-d_2}e^{-\frac{1}{2}\phi^{2}}d\phi$$\
Using (20),
$$= Ke^{-r\left(T-t\right)} N(-d_2)$$

Putting this all together from (23), we get that 
$$V_{EP}(S,t)= -G+H = -SN(-d_1) + Ke^{-r\left(T-t\right)} N(-d_2)$$\
$$\Large \implies \boxed{V_{EP}(S,t)= -SN(-d_1) + Ke^{-r\left(T-t\right)} N(-d_2)}$$
where
$$d_1 = \frac{\left(\log{S/K}+\left(r+\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}$$
and 
$$d_2 = \frac{\left(\log{S/K}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}=d_1 - \sigma\sqrt{T-t}$$

<h1><center> <ins><b>Formula for a Binary Call Option Value</b></ins></center></h1>

For a Binary Call, the payoff can be written by the function $$\text{Payoff}(e^{x'})=\begin{cases} 1 & \text{if } e^{x'} \ge K \iff x' \ge \log{K}\\ 0 & \text{if } e^{x'} < K \iff x' < \log{K}\end{cases}$$
Letting $V_{BC}(S,t)$ be the Binary Call Option Value of the underlying asset $S$ at time $t$ and using (18),
$$V_{BC}(S,t)=\frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}}\left( \int_{-\infty}^{\log{K}} 0 +\int_{\log{K}}^{\infty} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}\,dx' \right)$$
$$=\frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}}\int_{\log{K}}^{\infty} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}\,dx'$$
Looking earlier at our derivation of a European Call Option Value, we can see that our integral now is equivalent to $\frac{D}{K}$ in (19). Recalling our derivation of D above,
$$V_{BC}(S,t)= \frac{Ke^{-r\left(T-t\right)} N(d_2)}{K}$$
<br>
$$\Large \implies \boxed{V_{BC}(S,t)= e^{-r\left(T-t\right)} N(d_2)}$$
where
$$d_2 = \frac{\left(\log{S/K}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}$$

<h1><center> <ins><b>Formula for a Binary Put Option Value</b></ins></center></h1>

For a Binary Put, the payoff can be written by the function $$\text{Payoff}(e^{x'})=\begin{cases} 1 & \text{if } e^{x'} \le K \iff x' \le \log{K}\\ 0 & \text{if } e^{x'} > K \iff x' > \log{K}\end{cases}$$
Letting $V_{BP}(S,t)$ be the Binary Put Option Value of the underlying asset $S$ at time $t$ and using (18),
$$V_{BP}(S,t)=\frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}}\left( \int_{-\infty}^{\log{K}} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}\,dx' +\int_{\log{K}}^{\infty} 0 \right)$$
$$=\frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}}\int_{-\infty}^{\log{K}} e^{-\frac{\left(\log{S}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}\,dx'$$
Looking earlier at our derivation of a European Put Option Value, we can see that our integral now is equivalent to $\frac{H}{K}$ in (23). Recalling our derivation of H above,
$$V_{BP}(S,t)= \frac{Ke^{-r\left(T-t\right)} N(-d_2)}{K}$$
<br>
$$\Large \implies \boxed{V_{BP}(S,t)= e^{-r\left(T-t\right)} N(-d_2)}$$
where
$$d_2 = \frac{\left(\log{S/K}+\left(r-\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}$$

<h1><center> <ins><b>Factoring in Continuous Dividend Yield on the Underlying Asset</b></ins></center></h1>

The only difference in the derivation when you have a dividend-yield paying stock lies in the value of the Riskless Portfolio $\Pi$.\
The financial meaning here is the key: to delta-hedge your option you buy a quantity $\delta$ of the stock $S$, and only the stock is paying you the dividend, so you have to add this contribution in time to your hedge. The change in value of the stock is therefore:
$$dS = \left(\mu-q\right) Sdt + \sigma SdX$$
Let $\Pi$ denote the value of a portfolio of one long option position and a short position in
some quantity , $\Delta$, of the underlying:
$$\Pi = V(S,t) - \Delta S \tag{24}$$

It is natural to ask how the value of the portfolio changes from time t to t + dt. The change in
the portfolio value is due partly to the change in the option value and partly to the change in
the underlying:
$$d\Pi = dV - \Delta dS -q\Delta S dt$$

Notice that  has not changed during the time step; we have not anticipated the change in S.
From Ito we have:
$$dV = \frac{\partial V}{\partial t}dt + \frac{\partial V}{\partial S}dS + \frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}V}{\partial S^{2}}dt$$

Therefore using (1),
$$d\Pi = \frac{\partial V}{\partial t}dt + \frac{\partial V}{\partial S}dS + \frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}V}{\partial S^{2}}dt - \Delta dS -q\Delta Sdt \tag{25}$$

The right-hand side of (5.2) contains two types of terms, the deterministic and the random.
The deterministic terms are those with the dt, and the random terms are those with the dS.
Pretending for the moment that we know V and its derivatives then we know everything about
the right-hand side of (5.2) except for the value of dS. And this quantity we can never know in
advance.
These random terms are the risk in our portfolio. Is there any way to reduce or even eliminate
this risk? This can be done in theory (and almost in practice) by carefully choosing . The
random terms in (5.2) are
$$\left(\frac{\partial V}{\partial S} - \Delta\right) dS$$
And thus to reduce randomness to 0 (delta hedging), let $$\Delta = \frac{\partial V}{\partial S} \tag{26}$$

By the principle of **No Arbitrage**, due to the riskless change in our portfolio over a given time, the change (denoted as $d\Pi$ must be equivalent to the growth we would get if we put the equivalent amount
of cash in a risk-free interest-bearing account. This implies that
$$d\Pi = r \Pi dt \tag{27}$$

Substituting (24),(25),(26) into (27), we get
$$\left(\frac{\partial V}{\partial t} + \frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}V}{\partial S^{2}}-qS\frac{\partial V}{\partial S}\right)dt = r \left(V - S\frac{\partial V}{\partial S}\right)dt$$
And dividing by $dt$ and rearranging,
$$\Large \implies \boxed{\frac{\partial V}{\partial t} + \frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}V}{\partial S^{2}} + \left(r-q\right)S\frac{\partial V}{\partial S} - rV = 0}$$\
This is the derivation of the **Black-Scholes Equation** with a continuous dividend yield of $q$.

We can largely follow the same steps we used in **"Deriving the Black-Scholes Values for European and Binary Call/Put Options"** and we should end with
$$V(S,t) = \frac{e^{-r\left(T-t\right)}}{\sigma\,\sqrt{2\pi\left(T-t\right)}} \int_{-\infty}^{\infty} e^{-\frac{\left(\log{S}+\left(r-q-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)-x'\right)^{2}}{2\sigma^{2}\,\left(T-t\right)}}\text{Payoff}(e^{x'})\,dx'$$

And for finding the European Call and Put Option Values with continious dividend yield ($V_{EC_{D}}$ and $V_{EP_{D}}$), we follow the same steps used in those respective sections to get

$$ \Large \boxed{V_{EC_{D}}(S,t)=Se^{-q\,\left(T-t\right)} N(d_1) - Ke^{-r\,\left(T-t\right)} N(d_2)}$$\
$$ \Large \boxed{V_{EP_{D}}(S,t)= -Se^{-q\,\left(T-t\right)} N(-d_1) + Ke^{-r\left(T-t\right)} N(-d_2)}$$
where
$$d_1 = \frac{\left(\log{S/K}+\left(r-q+\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}$$
and 
$$d_2 = \frac{\left(\log{S/K}+\left(r-q-\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}}=d_1 - \sigma\sqrt{T-t}$$
