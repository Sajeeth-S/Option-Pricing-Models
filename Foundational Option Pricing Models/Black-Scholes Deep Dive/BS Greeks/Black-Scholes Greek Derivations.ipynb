<h1><center> <ins><b>Black-Scholes Greeks</b></ins></center></h1>

In this notebook, we will be deriving the **Black Scholes Greeks**. Namely, the derivatives with respect to the parameters of the European and Binary options under the Black-Scholes assumptions. They are the:
* **Delta**
* **Gamma**
* **Speed**
* **Theta**
* **Vega**
* **Rho**

We will find the Greeks of **European and Binary Calls/Puts**.

Before we begin, let us recall some formulae we derived for the option value of European and Binary Calls/Puts with continious dividend yields:\
<br>
$$\Large \boxed{V_{EC_{D}}(S,t)=Se^{-q\,\left(T-t\right)} N(d_1) - Ke^{-r\,\left(T-t\right)} N(d_2)}$$
<br>
$$\Large \boxed{V_{EP_{D}}(S,t)= -Se^{-q\,\left(T-t\right)} N(-d_1) + Ke^{-r\left(T-t\right)} N(-d_2)}$$
<br>
$$\Large \boxed{V_{BC}(S,t)= e^{-r\,\left(T-t\right)}N(d_2)}$$
<br>
$$\Large \boxed{V_{BP}(S,t)= e^{-r\,\left(T-t\right)} N(-d_2)}$$
Where,
$$\Large d_1 = \frac{\log{(S/K)}+\left(r-q+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{\sigma\,\sqrt{T-t}}$$
And,
$$\Large d_2 = \frac{\log{(S/K)}+\left(r-q-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{\sigma\,\sqrt{T-t}}= d_1 - \sigma\sqrt{T-t}$$
And let,
$$\Large N(x) = \frac{1}{\sqrt{2\pi}}\int_{-\infty}^{x} e^{-\frac{1}{2}\phi^{2}} d\phi$$
And,
$$\Large n(x) = N'(x)=\frac{1}{\sqrt{2\pi}}e^{-\frac{1}{2}x^{2}}$$

## <ins>Delta</ins>
The **delta**, $\Delta$, of an option is the sensitivity of the option to the underlying. It is the rate of change of value with respect to
the asset.$$$$
$$\Large \boxed{\Delta=\frac{\partial V}{\partial S}}$$

### Delta of European Call Option $(\Delta_{EC})$
$$\Delta_{EC}=\frac{\partial V_{EC_{D}}}{\partial S}=e^{-q\,\left(T-t\right)}N(d_1) + Se^{-q\left(T-t\right)}\frac{\partial N(d_1)}{\partial S} - Ke^{-r\,\left(T-t\right)}\frac{\partial N(d_2)}{\partial S}$$
$$=e^{-q\,\left(T-t\right)}N(d_1) + Se^{-q\left(T-t\right)}\frac{\partial N(d_1)}{\partial S} - Ke^{-r\,\left(T-t\right)}\frac{\partial N\left(d_1 - \sigma\sqrt{T-t}\right)}{\partial S}$$
$$=e^{-q\,\left(T-t\right)}N(d_1) + Se^{-q\left(T-t\right)}\frac{\partial (d_1)}{\partial S} n(d_1) - Ke^{-r\,\left(T-t\right)}\frac{\partial \left(d_1 - \sigma\sqrt{T-t}\right)}{\partial S} n\left(d_1-\sigma\sqrt{T-t}\right)$$
where $n(x) = N'(x)=\frac{1}{\sqrt{2\pi}}e^{-\frac{1}{2}x^{2}}$
$$\Delta_{EC} = e^{-q\,\left(T-t\right)}N(d_1) + \frac{Se^{-q\left(T-t\right)}}{\sigma S\sqrt{T-t}}\,n(d_1) - \frac{Ke^{-r\,\left(T-t\right)}}{\sigma S\sqrt{T-t}}\,n\left(d_1-\sigma\sqrt{T-t}\right)$$
<br>
Let us inspect $n\left(d_1-\sigma\sqrt{T-t}\right)$ closer,
$$n\left(d_1-\sigma\sqrt{T-t}\right)=\frac{1}{\sqrt{2\pi}}e^{-\frac{1}{2}\left(d_1-\sigma\sqrt{T-t}\right)^{2}}$$
$$=\frac{1}{\sqrt{2\pi}}e^{-\frac{1}{2}\left(d_1\right)^{2}}\times e^{d_1 \sigma\sqrt{T-t}-\frac{1}{2}\sigma^{2}\left(T-t\right)} $$
$$=n(d_1)\,e^{-\frac{1}{2}\sigma^{2}\left(T-t\right) + \frac{\left(\log{S/K}+\left(r-q+\frac{1}{2}\sigma^{2}\right)\left(T-t\right) \right) }{\sigma\,\sqrt{T-t}} \sigma\sqrt{T-t}}$$ 
$$=n(d_1)\,e^{-\frac{1}{2}\sigma^{2}\left(T-t\right) + \log{S/K}+\left(r-q+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}$$
$$=n(d_1)\,e^{\log{S/K}}e^{\left(r-q\right)\left(T-t\right)}$$
$$=n(d_1)\,\frac{S}{K}e^{\left(r-q\right)\left(T-t\right)}\tag{1}$$
Substituting this into $\Delta_{EC}$ above,
$$\Delta_{EC} = e^{-q\,\left(T-t\right)}N(d_1) + \frac{Se^{-q\left(T-t\right)}}{\sigma S\sqrt{T-t}}\,n(d_1) - \frac{Ke^{-r\,\left(T-t\right)}}{\sigma S\sqrt{T-t}}\frac{S}{K}e^{\left(r-q\right)\left(T-t\right)}\,n(d_1)$$
$$=e^{-q\,\left(T-t\right)}N(d_1) + \frac{e^{-q\left(T-t\right)}}{\sigma \sqrt{T-t}}\,n(d_1) - \frac{e^{-q\left(T-t\right)}}{\sigma \sqrt{T-t}}\,n(d_1)$$\
$$\Large \implies \boxed{\Delta_{EC}=e^{-q\,\left(T-t\right)}N(d_1)}\tag{2}$$

### Delta of European Put Option $(\Delta_{EP})$
$$\Delta_{EP}=\frac{\partial V_{EP_{D}}}{\partial S}=-e^{-q\,\left(T-t\right)}N(-d_1) - Se^{-q\,\left(T-t\right)}\frac{\partial N(-d_1)}{\partial S} + Ke^{-r\,\left(T-t\right)}\frac{\partial N(-d_2)}{\partial S}$$
$$=-e^{-q\,\left(T-t\right)}N(-d_1) - Se^{-q\,\left(T-t\right)}\frac{\partial N(-d_1)}{\partial S} + Ke^{-r\,\left(T-t\right)}\frac{\partial N\left(-d_1 + \sigma\sqrt{T-t}\right)}{\partial S}$$
$$=-e^{-q\,\left(T-t\right)}N(-d_1) - Se^{-q\,\left(T-t\right)}\frac{\partial (-d_1)}{\partial S} n(-d_1) + Ke^{-r\,\left(T-t\right)}\frac{\partial \left(-d_1 + \sigma\sqrt{T-t}\right)}{\partial S} n\left(-d_1+\sigma\sqrt{T-t}\right)$$
where $n(x) = N'(x)=\frac{1}{\sqrt{2\pi}}e^{-\frac{1}{2}x^{2}}$. Also, note that clearly $n(-x)=n(x)$ due to the $x^{2}$ inside the function. So,
$$\Delta_{EP} = -e^{-q\,\left(T-t\right)}N(-d_1) - \frac{Se^{-q\,\left(T-t\right)}}{\sigma S\sqrt{T-t}}\,n(d_1) + \frac{Ke^{-r\,\left(T-t\right)}}{\sigma S\sqrt{T-t}}\,n\left(d_1-\sigma\sqrt{T-t}\right)$$
From (1), we know $$ n\left(d_1-\sigma\sqrt{T-t}\right)=n(d_1)\,\frac{S}{K}e^{\left(r-q\right)\,\left(T-t\right)}$$,
$$\implies \Delta_{EP} = -e^{-q\,\left(T-t\right)}N(-d_1) - \frac{Se^{-q\,\left(T-t\right)}}{\sigma S\sqrt{T-t}}\,n(d_1) + \frac{Ke^{-r\,\left(T-t\right)}}{\sigma S\sqrt{T-t}}\frac{S}{K}e^{\left(r-q\right)\,\left(T-t\right)}\,n(d_1)$$
$$=-e^{-q\,\left(T-t\right)}N(-d_1) - \frac{e^{-q\,\left(T-t\right)}}{\sigma \sqrt{T-t}}\,n(d_1) + \frac{e^{-q\,\left(T-t\right)}}{\sigma \sqrt{T-t}}\,n(d_1)$$\
Using the fact that $N(-x)=1-N(x)$,
$$\implies \Delta_{EP}=-e^{-q\,\left(T-t\right)}\left(1-N(d_1)\right)$$
Taking the negative sign into the bracket,
$$\Large \boxed{\Delta_{EP}=e^{-q\,\left(T-t\right)}\left(N(d_1)-1\right)}\tag{3}$$

### Delta of Binary Call Option $(\Delta_{BC})$
$$\Delta_{BC}=\frac{\partial V_{BC}}{\partial S}=e^{-r\,\left(T-t\right)}\frac{\partial N(d_2)}{\partial S}$$
$$=e^{-r\,\left(T-t\right)}n(d_2)\frac{\partial (d_2)}{\partial S}$$
where $n(x) = N'(x)=\frac{1}{\sqrt{2\pi}}e^{-\frac{1}{2}x^{2}}$
So clearly,
$$\Delta_{BC}=e^{-r\,\left(T-t\right)}n(d_2)\frac{1}{\sigma S\sqrt{T-t}}$$\
$$\Large \implies \boxed{\Delta_{BC}=e^{-r\,\left(T-t\right)}\frac{n(d_2)}{\sigma S\sqrt{T-t}}}\tag{4}$$

### Delta of Binary Put Option $(\Delta_{BP})$
$$\Delta_{BP}=\frac{\partial V_{BP}}{\partial S}=e^{-r\,\left(T-t\right)}\frac{\partial N(-d_2)}{\partial S}$$
$$=e^{-r\,\left(T-t\right)}n(-d_2)\frac{\partial (-d_2)}{\partial S}$$
where $n(x) = N'(x)=\frac{1}{\sqrt{2\pi}}e^{-\frac{1}{2}x^{2}}$
So clearly,
$$\Delta_{BP}=e^{-r\,\left(T-t\right)}n(d_2)\frac{-1}{\sigma S\sqrt{T-t}}$$\
$$\Large \implies \boxed{\Delta_{BP}=-e^{-r\,\left(T-t\right)}\frac{n(d_2)}{\sigma S\sqrt{T-t}}}\tag{5}$$

## <ins>Gamma</ins>
The **gamma**, $\Gamma$, of an option is the sensitivity of the delta to the underlying. It is a measure of by how much or how often a position must be rehedged in order to maintain a delta-neutral position. It is the second derivative of the option value with respect to the underlying asset.$$$$
$$\Large \boxed{\Gamma=\frac{\partial^{2} V}{\partial S^{2}}}$$

### Gamma of European Call Option $(\Gamma_{EC})$
$$\Gamma_{EC}=\frac{\partial^{2} V_{EC_{D}}}{\partial S^{2}}=\frac{\partial}{\partial S}\left( \frac{\partial V_{EC_{D}}}{\partial S}\right)=\frac{\partial}{\partial S}\Delta_{EC}$$\
Substituing the value of $\Delta_{EC}$ we derived earlier in (2),\
$$\Gamma_{EC}=\frac{\partial}{\partial S}\left(e^{-q\,\left(T-t\right)}N(d_1)\right)$$
$$=e^{-q\,\left(T-t\right)}\frac{\partial (d_1)}{\partial S}n(d_1)$$
where $n(x) = N'(x)=\frac{1}{\sqrt{2\pi}}e^{-\frac{1}{2}x^{2}}$. And so,
$$\Gamma_{EC}=e^{-q\,\left(T-t\right)}\frac{1}{\sigma S\sqrt{T-t}}n(d_1)$$\
$$\Large \implies \boxed{\Gamma_{EC}=\frac{e^{-q\,\left(T-t\right)}n(d_1)}{\sigma S\sqrt{T-t}}}\tag{6}$$

### Gamma of European Put Option $(\Gamma_{EP})$
$$\Gamma_{EP}=\frac{\partial^{2} V_{EP_{D}}}{\partial S^{2}}=\frac{\partial}{\partial S}\left( \frac{\partial V_{EP_{D}}}{\partial S}\right)=\frac{\partial}{\partial S}\Delta_{EP}$$\
Substituing the value of $\Delta_{EP}$ we derived earlier in (3),\
$$\Gamma_{EP}=\frac{\partial}{\partial S}\left(e^{-q\,\left(T-t\right)}\left(N(d_1)-1\right)\right)$$
$$=\frac{\partial}{\partial S}\left(e^{-q\,\left(T-t\right)}N(d_1)\right)-\frac{\partial}{\partial S}\left(e^{-q\,\left(T-t\right)}\right) $$
$$=\Gamma_{EC} - 0 $$
Using our earlier derived value of $\Gamma_{EC}$ in (6),$$$$
$$\Large \implies \boxed{\Gamma_{EP}=\frac{e^{-q\,\left(T-t\right)}n(d_1)}{\sigma S\sqrt{T-t}}}$$

### Gamma of Binary Call Option $(\Gamma_{BC})$
$$\Gamma_{BC}=\frac{\partial^{2} V_{BC}}{\partial S^{2}}=\frac{\partial}{\partial S}\left( \frac{\partial V_{BC}}{\partial S}\right)=\frac{\partial}{\partial S}\Delta_{BC}$$\
Substituing the value of $\Delta_{BC}$ we derived earlier in (4),\
$$\Gamma_{BC}=\frac{\partial}{\partial S}\left(e^{-r\,\left(T-t\right)}\frac{n(d_2)}{\sigma S\sqrt{T-t}}\right)$$
$$=\frac{e^{-r\,\left(T-t\right)}}{\sigma \sqrt{T-t}}\frac{\partial}{\partial S}\left(\frac{n(d_2)}{S}\right)$$
where $n(x) = N'(x)=\frac{1}{\sqrt{2\pi}}e^{-\frac{1}{2}x^{2}}$. And so,
$$\Gamma_{BC}=\frac{e^{-r\,\left(T-t\right)}}{\sigma \sqrt{T-t}} \left[ \frac{1}{S}\frac{\partial}{\partial S}\left(n(d_2)\right) + n(d_2)\frac{\partial}{\partial S}\left(\frac{1}{S}\right) \right]$$
$$=\frac{e^{-r\,\left(T-t\right)}}{\sigma \sqrt{T-t}} \left[ \frac{1}{S}\frac{\partial}{\partial S}\left( \frac{1}{\sqrt{2\pi}}e^{-\frac{1}{2}(d_2)^{2}}\right) - \frac{n(d_2)}{S^2} \right]$$
$$=\frac{e^{-r\,\left(T-t\right)}}{\sigma \sqrt{T-t}}\left[ \frac{1}{S}\frac{\partial}{\partial S}\left( -\frac{1}{2}\left(d_2\right)^{2}\right)\left( \frac{1}{\sqrt{2\pi}}e^{-\frac{1}{2}(d_2)^{2}}\right) - \frac{n(d_2)}{S^2} \right]$$
$$=\frac{e^{-r\,\left(T-t\right)}}{\sigma \sqrt{T-t}}\left[ -\frac{d_2}{S}\frac{\partial}{\partial S}\left(d_2\right)n(d_2) - \frac{n(d_2)}{S^2} \right]$$
$$=\frac{e^{-r\,\left(T-t\right)}}{\sigma \sqrt{T-t}}\left[ -\frac{d_2}{\sigma S^{2}\sqrt{T-t}}n(d_2) - \frac{n(d_2)}{S^2} \right]$$
$$=-\frac{e^{-r\,\left(T-t\right)}n(d_2)}{\sigma^{2}S^{2} \left(T-t\right)}\left[ d_2 - \sigma \sqrt{T-t} \right]$$\
$$\Large \implies \boxed{\Gamma_{BC}=-\frac{e^{-r\,\left(T-t\right)}d_1n(d_2)}{\sigma^{2}S^{2} \left(T-t\right)}}\tag{7}$$

### Gamma of Binary Put Option $(\Gamma_{BP})$
$$\Gamma_{BP}=\frac{\partial^{2} V_{BP}}{\partial S^{2}}=\frac{\partial}{\partial S}\left( \frac{\partial V_{BP}}{\partial S}\right)=\frac{\partial}{\partial S}\Delta_{BP}$$\
Substituing the value of $\Delta_{BP}$ we derived earlier in (5),\
$$\Gamma_{BP}=\frac{\partial}{\partial S}\left(-e^{-r\,\left(T-t\right)}\frac{n(d_2)}{\sigma S\sqrt{T-t}}\right)$$
$$=-\frac{\partial}{\partial S}\left(e^{-r\,\left(T-t\right)}\frac{n(d_2)}{\sigma S\sqrt{T-t}}\right)$$
Using (7) we can say,
$$\Gamma_{BP}=-\Gamma_{BC}$$\
$$\Large \implies \boxed{\Gamma_{BP}=\frac{e^{-r\,\left(T-t\right)}d_1n(d_2)}{\sigma^{2}S^{2} \left(T-t\right)}}$$

## <ins>Speed</ins>
The **speed** of an option is the rate of change of the gamma with respect to the underlying. It is important for higher order Taylor Series expansions.$$$$
$$\Large \boxed{\text{Speed}=\frac{\partial^{3} V}{\partial S^{3}}}$$

### Speed of European Call Option $(\text{Speed}_{EC})$
$$\text{Speed}_{EC}=\frac{\partial^{3} V_{EC_{D}}}{\partial S^{3}}=\frac{\partial}{\partial S}\left( \frac{\partial^{2} V_{EC_{D}}}{\partial S^{2}}\right)=\frac{\partial}{\partial S}\Gamma_{EC}$$\
Substituing the value of $\Gamma_{EC}$ we derived earlier in (6),\
$$\text{Speed}_{EC}=\frac{\partial}{\partial S}\left(\frac{e^{-q\,\left(T-t\right)}n(d_1)}{\sigma S\sqrt{T-t}}\right)$$
$$=\frac{e^{-q\,\left(T-t\right)}}{\sigma \sqrt{T-t}}\frac{\partial}{\partial S}\left(\frac{n(d_1)}{S}\right)$$
Using our derivation of $\frac{\partial}{\partial S}\left(\frac{n(d_2)}{S}\right)$ earlier in (x) and the fact that $\frac{\partial (d_1)}{\partial S}=\frac{\partial (d_2)}{\partial S}$,
$$\text{Speed}_{EC} =\frac{e^{-q\,\left(T-t\right)}}{\sigma \sqrt{T-t}}\left[ -\frac{d_1}{\sigma S^{2}\sqrt{T-t}}n(d_1) - \frac{n(d_1)}{S^2} \right]$$\
$$\Large \implies \boxed{\text{Speed}_{EC} =-\frac{e^{-q\,\left(T-t\right)}n(d_1)}{\sigma^{2}S^{2} \left(T-t\right)}\left( d_1+ \sigma \sqrt{T-t}\right)}$$

### Speed of European Put Option $(\text{Speed}_{EP})$
$$\text{Speed}_{EP}=\frac{\partial^{3} V_{EP_{D}}}{\partial S^{3}}=\frac{\partial}{\partial S}\left( \frac{\partial^{2} V_{EP_{D}}}{\partial S^{2}}\right)=\frac{\partial}{\partial S}\Gamma_{EP}$$\
Recalling from our derivation of the Gammas that $\Gamma_{EC}=\Gamma_{EP}$,\
$$\text{Speed}_{EP}=\frac{\partial}{\partial S}\Gamma_{EC}$$
Using the formula above,
$$\Large \implies \boxed{\text{Speed}_{EP} =-\frac{e^{-q\,\left(T-t\right)}n(d_1)}{\sigma^{2}S^{2} \left(T-t\right)}\left( d_1+ \sigma \sqrt{T-t}\right)}$$

### Speed of Binary Call Option $(\text{Speed}_{BC})$
$$\text{Speed}_{BC}=\frac{\partial^{3} V_{BC}}{\partial S^{3}}=\frac{\partial}{\partial S}\left( \frac{\partial^{2} V_{BC}}{\partial S^{2}}\right)=\frac{\partial}{\partial S}\Gamma_{BC}$$\
Substituing the value of $\Gamma_{BC}$ we derived earlier in (7),\
$$\text{Speed}_{BC}=\frac{\partial}{\partial S}\left[ -\frac{e^{-r\,\left(T-t\right)}d_1n(d_2)}{\sigma^{2}S^{2} \left(T-t\right)} \right]$$
$$=-\frac{e^{-r\left(T-t\right)}}{\sigma^{2}\left(T-t\right)} \frac{\partial}{\partial S}\left(\frac{d_1 n(d_2)}{S^{2}}\right)$$
$$=-\frac{e^{-r\left(T-t\right)}}{\sigma^{2}\left(T-t\right)} \frac{\partial}{\partial S}\left(\frac{d_1}{S}\frac{n(d_2)}{S}\right)$$
$$=-\frac{e^{-r\left(T-t\right)}}{\sigma^{2}\left(T-t\right)}\left[\frac{\partial}{\partial S}\left(\frac{d_1}{S}\right)\frac{n(d_2)}{S}+\frac{d_1}{S}\frac{\partial}{\partial S}\left(\frac{n(d_2)}{S}\right)\right] $$
$$=-\frac{e^{-r\left(T-t\right)}}{\sigma^{2}\left(T-t\right)}\left\{\left[\frac{1}{S}\frac{\partial}{\partial S}(d_1)+d_1\frac{\partial}{\partial S}\left(\frac{1}{S}\right)\right]\frac{n(d_2)}{S} + \frac{d_1}{S}\left[ -\frac{d_2}{\sigma S^{2}\sqrt{T-t}}n(d_2) - \frac{n(d_2)}{S^2} \right]\right\} $$
$$=-\frac{e^{-r\left(T-t\right)}}{\sigma^{2}\left(T-t\right)}\left[\left( \frac{1}{\sigma S^{2}\sqrt{T-t}}-\frac{d_1}{S^2}\right)\frac{n(d_2)}{S} + \frac{d_1}{S}\left( -\frac{d_2}{\sigma S^{2}\sqrt{T-t}}n(d_2) - \frac{n(d_2)}{S^2} \right)\right]$$
$$=-\frac{e^{-r\left(T-t\right)}}{\sigma^{2}\left(T-t\right)}\left[\frac{n(d_2)}{S^{3}}\left( \frac{1}{\sigma \sqrt{T-t}}-d_1\right) + \frac{n(d_2)}{S^{3}}\left( -\frac{d_1d_2}{\sigma \sqrt{T-t}} -d_1 \right)\right] $$\
$$\Large \implies \boxed{\text{Speed}_{BC}=-\frac{e^{-r\,\left(T-t\right)}n(d_2)}{\sigma^{2}S^{3}\left(T-t\right)}\left( \frac{1-d_1d_2}{\sigma \sqrt{T-t}}-2d_1 \right)} $$

### Speed of Binary Put Option $(\text{Speed}_{BP})$
$$\text{Speed}_{BP}=\frac{\partial^{3} V_{BP}}{\partial S^{3}}=\frac{\partial}{\partial S}\left( \frac{\partial^{2} V_{BP}}{\partial S^{2}}\right)=\frac{\partial}{\partial S}\Gamma_{BP}$$
Recalling from our Gamma derivations that $\Gamma_{BC}=-\Gamma_{BP}$,
$$\text{Speed}_{BP}=-\frac{\partial}{\partial S}\Gamma_{BC}=-\text{Speed}_{BC}$$
And using the value of $\text{Speed}_{BC}$ we found earlier,$$$$
$$\Large \implies \boxed{\text{Speed}_{BP}=\frac{e^{-r\,\left(T-t\right)}n(d_2)}{\sigma^{2}S^{3}\left(T-t\right)}\left( \frac{1-d_1d_2}{\sigma \sqrt{T-t}}-2d_1 \right)} $$

## <ins>Theta</ins>
The **theta**,$\,\Theta$, of an option is the rate of change of the option value with respect to the time.  In a delta-hedged portfolio the theta contributes to ensuring that the portfolio earns the risk-free rate.$$$$
$$\Large \boxed{\Theta=\frac{\partial V}{\partial t}}$$

### Theta of European Call Option $(\Theta_{EC})$
$$\Theta_{EC}=\frac{\partial V_{EC_{D}}}{\partial t}=\frac{\partial}{\partial t}\left(Se^{-q\,\left(T-t\right)} N(d_1) - Ke^{-r\,\left(T-t\right)} N(d_2)\right)$$
$$=S\left(qe^{-q\,\left(T-t\right)} N(d_1)+e^{-q\,\left(T-t\right)}\frac{\partial}{\partial t}\left(N(d_1)\right)\right) - K\left( re^{-r\,\left(T-t\right)} N(d_2)+e^{-r\,\left(T-t\right)}\frac{\partial}{\partial t}\left(N(d_2)\right)\right)$$
$$=S\left(qe^{-q\,\left(T-t\right)} N(d_1)+e^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial t}\left(d_1\right)\right) - K\left( re^{-r\,\left(T-t\right)} N(d_2)+e^{-r\,\left(T-t\right)}n(d_2)\frac{\partial}{\partial t}\left(d_2\right)\right)$$
Where $n(x) = N'(x)=\frac{1}{\sqrt{2\pi}}e^{-\frac{1}{2}x^{2}}$$$$$
$$\Theta_{EC}=S\left(qe^{-q\,\left(T-t\right)} N(d_1)+e^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial t}\left(d_1\right)\right) - K\left( re^{-r\,\left(T-t\right)} N(d_2)+e^{-r\,\left(T-t\right)}n(d_2)\frac{\partial}{\partial t}\left(d_1-\sigma\sqrt{T-t}\right)\right)$$
Since $d_2=d_1-\sigma \sqrt{T-t}$.$$$$
$$\Theta_{EC}=S\left(qe^{-q\,\left(T-t\right)} N(d_1)+e^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial t}\left(d_1\right)\right) - K\left( re^{-r\,\left(T-t\right)} N(d_2)+e^{-r\,\left(T-t\right)}n(d_2)\left[\frac{\partial}{\partial t}\left(d_1\right)+\frac{\sigma}{2\sqrt{T-t}}\right]\right)$$
$$$$
Using (1) to put $n(d_2)$ into terms of $n(d_1)$
$$\Theta_{EC}=S\left(qe^{-q\,\left(T-t\right)} N(d_1)+e^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial t}\left(d_1\right)\right) - K\left( re^{-r\,\left(T-t\right)} N(d_2)+e^{-r\,\left(T-t\right)}\left[n(d_1)\,\frac{S}{K}e^{\left(r-q\right)\left(T-t\right)}\right]\left[\frac{\partial}{\partial t}\left(d_1\right)+\frac{\sigma}{2\sqrt{T-t}}\right]\right)$$$$$$
Expanding and collecting terms,
$$\Theta_{EC}=qSN(d_1)e^{-q\,\left(T-t\right)} -rKN(d_2)e^{-r\,\left(T-t\right)} +Sn(d_1)e^{-q\,\left(T-t\right)}\frac{\partial}{\partial t}\left(d_1\right)-Sn(d_1)e^{-q\,\left(T-t\right)}\frac{\partial}{\partial t}\left(d_1\right)-\frac{\sigma Sn(d_1)e^{-q\,\left(T-t\right)}}{2\sqrt{T-t}}$$$$$$
$$\Large \implies \boxed{\Theta_{EC}=qSN(d_1)e^{-q\,\left(T-t\right)} -rKN(d_2)e^{-r\,\left(T-t\right)}-\frac{\sigma S\,n(d_1)e^{-q\,\left(T-t\right)}}{2\sqrt{T-t}}}$$

### Theta of European Put Option $(\Theta_{EP})$
$$\Theta_{EP}=\frac{\partial V_{EP_{D}}}{\partial t}=\frac{\partial}{\partial t}\left(-Se^{-q\,\left(T-t\right)} N(-d_1) + Ke^{-r\,\left(T-t\right)} N(-d_2)\right)$$
$$=-S\left(qe^{-q\,\left(T-t\right)} N(-d_1)+e^{-q\,\left(T-t\right)}\frac{\partial}{\partial t}\left(N(-d_1)\right)\right) + K\left( re^{-r\,\left(T-t\right)} N(-d_2)+e^{-r\,\left(T-t\right)}\frac{\partial}{\partial t}\left(N(-d_2)\right)\right)$$
$$=-S\left(qe^{-q\,\left(T-t\right)} N(-d_1)+e^{-q\,\left(T-t\right)}n(-d_1)\frac{\partial}{\partial t}\left(-d_1\right)\right) + K\left( re^{-r\,\left(T-t\right)} N(-d_2)+e^{-r\,\left(T-t\right)}n(-d_2)\frac{\partial}{\partial t}\left(-d_2\right)\right)$$
Where $n(x) = N'(x)=\frac{1}{\sqrt{2\pi}}e^{-\frac{1}{2}x^{2}}$$$$$
$$\Theta_{EP}=-S\left(qe^{-q\,\left(T-t\right)} N(-d_1)+e^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial t}\left(-d_1\right)\right) + K\left( re^{-r\,\left(T-t\right)} N(-d_2)+e^{-r\,\left(T-t\right)}n(d_2)\frac{\partial}{\partial t}\left(-d_1+\sigma\sqrt{T-t}\right)\right)$$
Since $d_2=d_1-\sigma \sqrt{T-t}$ and $n(-x)=n(x)$.$$$$
$$\Theta_{EP}=-S\left(qe^{-q\,\left(T-t\right)} N(-d_1)+e^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial t}\left(-d_1\right)\right) + K\left( re^{-r\,\left(T-t\right)} N(-d_2)+e^{-r\,\left(T-t\right)}n(d_2)\left[\frac{\partial}{\partial t}\left(-d_1\right)-\frac{\sigma}{2\sqrt{T-t}}\right]\right)$$
$$$$
Using (1) to put $n(d_2)$ into terms of $n(d_1)$
$$\Theta_{EP}=-S\left(qe^{-q\,\left(T-t\right)} N(-d_1)+e^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial t}\left(-d_1\right)\right) + K\left( re^{-r\,\left(T-t\right)} N(-d_2)+e^{-r\,\left(T-t\right)}\left[n(d_1)\,\frac{S}{K}e^{\left(r-q\right)\left(T-t\right)}\right]\left[\frac{\partial}{\partial t}\left(-d_1\right)-\frac{\sigma}{2\sqrt{T-t}}\right]\right)$$$$$$
Expanding and collecting terms,
$$\Theta_{EP}=-qSN(-d_1)e^{-q\,\left(T-t\right)} +rKN(-d_2)e^{-r\,\left(T-t\right)} -Sn(d_1)e^{-q\,\left(T-t\right)}\frac{\partial}{\partial t}\left(d_1\right)+Sn(d_1)e^{-q\,\left(T-t\right)}\frac{\partial}{\partial t}\left(d_1\right)-\frac{\sigma Sn(d_1)e^{-q\,\left(T-t\right)}}{2\sqrt{T-t}}$$$$$$
$$\Large \implies \boxed{\Theta_{EC}=-qSN(-d_1)e^{-q\,\left(T-t\right)} +rKN(-d_2)e^{-r\,\left(T-t\right)}-\frac{\sigma S\,n(d_1)e^{-q\,\left(T-t\right)}}{2\sqrt{T-t}}}$$

### Theta of Binary Call Option $(\Theta_{BC})$
$$\Theta_{BC}=\frac{\partial V_{BC}}{\partial t}=\frac{\partial}{\partial t}\left(e^{-r\,\left(T-t\right)} N(d_2)\right)$$
$$=re^{-r\,\left(T-t\right)} N(d_2)+e^{-r\,\left(T-t\right)} n(d_2)\frac{\partial}{\partial t}\left(d_2\right)\tag{8}$$
Where $n(x) = N'(x)=\frac{1}{\sqrt{2\pi}}e^{-\frac{1}{2}x^{2}}$$$$$
Let us look at $\frac{\partial}{\partial t}\left(d_2\right)$ closer using the Quotient Rule,
$$\frac{\partial}{\partial t}\left(d_2\right)=\frac{ -\left(r-q-\frac{1}{2}\sigma^{2}\right)\sigma \sqrt{T-t}-\left[ \log{S/K}+\left(r-q-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)\right]\left(\frac{-\sigma}{2\sqrt{T-t}}\right)}{\sigma^{2}\left(T-t\right)}$$
$$=-\frac{\left(r-q-\frac{1}{2}\sigma^{2}\right)}{\sigma \sqrt{T-t}}+\frac{log{S/K}+\left(r-q-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{2\sigma \sqrt{T-t}\left(T-t\right)} $$
$$=-\frac{\left(r-q\right)}{\sigma \sqrt{T-t}}+\frac{log{S/K}+\left(r-q-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{2\sigma \sqrt{T-t}\left(T-t\right)}+\frac{\frac{1}{2}\sigma^{2}}{\sigma \sqrt{T-t}}$$
$$=-\frac{\left(r-q\right)}{\sigma \sqrt{T-t}}+\frac{log{S/K}+\left(r-q-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{2\sigma \sqrt{T-t}\left(T-t\right)}+\frac{\sigma^{2}\left(T-t\right)}{2\sigma \sqrt{T-t}\left(T-t\right)}$$
$$=-\frac{\left(r-q\right)}{\sigma \sqrt{T-t}}+\frac{log{S/K}+\left(r-q+\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{2\sigma \sqrt{T-t}\left(T-t\right)}$$\
$$=\frac{d_1}{2\left(T-t\right)}-\frac{\left(r-q\right)}{\sigma \sqrt{T-t}}\tag{9}$$
Substiuting this back into (8),$$$$
$$\Large \implies \boxed{\Theta_{BC}=re^{-r\,\left(T-t\right)} N(d_2)+e^{-r\,\left(T-t\right)} n(d_2)\left(\frac{d_1}{2\left(T-t\right)}-\frac{r-q}{\sigma \sqrt{T-t}}\right)}$$

### Theta of Binary Put Option $(\Theta_{BP})$
$$\Theta_{BP}=\frac{\partial V_{BP}}{\partial t}=\frac{\partial}{\partial t}\left(e^{-r\,\left(T-t\right)} N(-d_2)\right)$$
$$=re^{-r\,\left(T-t\right)} N(-d_2)+e^{-r\,\left(T-t\right)} n(-d_2)\frac{\partial}{\partial t}\left(-d_2\right)$$
$$=re^{-r\,\left(T-t\right)} N(-d_2)-e^{-r\,\left(T-t\right)} n(d_2)\frac{\partial}{\partial t}\left(d_2\right)\tag{10}$$
Where $n(x) = N'(x)=\frac{1}{\sqrt{2\pi}}e^{-\frac{1}{2}x^{2}}$ and so $n(-x)=n(x)$$$$$
Recalling from (9) our derived value of $\frac{\partial}{\partial t}\left(d_2\right)$,$$$$
$$\frac{\partial}{\partial t}\left(d_2\right)=\frac{d_1}{2\left(T-t\right)}-\frac{\left(r-q\right)}{\sigma \sqrt{T-t}}$$
Substiuting this back into (10),$$$$
$$\Large \implies \boxed{\Theta_{BP}=re^{-r\,\left(T-t\right)} N(-d_2)-e^{-r\,\left(T-t\right)} n(d_2)\left(\frac{d_1}{2\left(T-t\right)}-\frac{r-q}{\sigma \sqrt{T-t}}\right)}$$

## <ins>Vega</ins>
The **vega** of an option is the rate of change of the option value with respect to the volatility. This is inaccurate in practice since it is extremely difficult to know volatility with certainty at any time, including the future.\
There is a downside to the measurement of vega. It is only really meaningful for options having single-signed gamma everywhere. For example it makes sense to measure vega for European calls and puts but not Binary calls and binary puts.$$$$
$$\Large \boxed{\text{Vega}=\frac{\partial V}{\partial \sigma}}$$

### Vega of European Call Option $(\text{Vega}_{EC})$
$$\text{Vega}_{EC}=\frac{\partial V_{EC_{D}}}{\partial \sigma}=\frac{\partial}{\partial \sigma}\left(Se^{-q\,\left(T-t\right)} N(d_1) - Ke^{-r\,\left(T-t\right)} N(d_2)\right)$$
$$=Se^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial \sigma}(d_1)-Ke^{-r\,\left(T-t\right)}n(d_2)\frac{\partial}{\partial \sigma}(d_2)\tag{11}$$
Now let us use the substitution in (1) to write $n(d_2)$ in terms of $n(d_1)$.$$$$
$$\text{Vega}_{EC}=Se^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial \sigma}(d_1)-Ke^{-r\,\left(T-t\right)}\frac{S}{K}n(d_1)e^{\left(r-q\right)\left(T-t\right)}\frac{\partial}{\partial \sigma}(d_1-\sigma \sqrt{T-t})$$
$$=Se^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial \sigma}(d_1)-Se^{-q\,\left(T-t\right)}n(d_1)\left[\frac{\partial}{\partial \sigma}(d_1)\,\sqrt{T-t}\right]$$
$$=Se^{-q\,\left(T-t\right)}n(d_1)\left[\frac{\partial}{\partial \sigma}(d_1)-\frac{\partial}{\partial \sigma}(d_1)+\sqrt{T-t}\right]$$\
$$\Large \implies \boxed{\text{Vega}_{EC}=Sn(d_1)\sqrt{T-t}\,e^{-q\,\left(T-t\right)}}$$

### Vega of European Put Option $(\text{Vega}_{EP})$
$$\text{Vega}_{EP}=\frac{\partial V_{EP_{D}}}{\partial \sigma}=\frac{\partial}{\partial \sigma}\left(-Se^{-q\,\left(T-t\right)} N(-d_1) + Ke^{-r\,\left(T-t\right)} N(-d_2)\right)$$
$$=-Se^{-q\,\left(T-t\right)}n(-d_1)\frac{\partial}{\partial \sigma}(-d_1)+Ke^{-r\,\left(T-t\right)}n(-d_2)\frac{\partial}{\partial \sigma}(-d_2)$$
$$=Se^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial \sigma}(d_1)-Ke^{-r\,\left(T-t\right)}n(d_2)\frac{\partial}{\partial \sigma}(d_2)$$
Since $n(-x)=n(x)$ and bringing the negative sign from inside to outside the derivative. Now using the work above, from (11), it is clear that
$$\text{Vega}_{EP}=\text{Vega}_{EC}$$\
$$\Large \implies \boxed{\text{Vega}_{EP}=Sn(d_1)\sqrt{T-t}\,e^{-q\,\left(T-t\right)}}$$

### Vega of Binary Call Option $(\text{Vega}_{BC})$
$$\text{Vega}_{BC}=\frac{\partial V_{BC}}{\partial \sigma}=\frac{\partial}{\partial \sigma}\left(e^{-r\,\left(T-t\right)} N(d_2)\right)$$
$$=e^{-r\,\left(T-t\right)}n(d_2)\frac{\partial}{\partial \sigma}\left(d_2\right)\tag{12}$$
Let us look at $\frac{\partial}{\partial \sigma}\left(d_2\right)$ closer and differentiate via Quotient Rule,
$$\frac{\partial}{\partial \sigma}\left(d_2\right)=\frac{\partial}{\partial \sigma}\left( \frac{\log{S/K}+\left(r-q-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{\sigma \sqrt{T-t}}\right)$$
$$=\frac{-\sigma\left(T-t\right)\sigma \sqrt{T-t}-\sqrt{T-t}\,\left[\log{S/K}+\left(r-q-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)\right]}{\sigma^{2}\left(T-t\right)}$$
$$=-\sqrt{T-t}-\frac{log{S/K}+\left(r-q-\frac{1}{2}\sigma^{2}\right)\left(T-t\right)}{\sigma \sqrt{T-t}}\frac{1}{\sigma}$$
$$=-\left(\sqrt{T-t}+\frac{d_2}{\sigma}\right)$$
Thus putting this into (12),$$$$
$$\Large \implies \boxed{\text{Vega}_{BC}=-e^{-r\,\left(T-t\right)}n(d_2)\left(\sqrt{T-t}+\frac{d_2}{\sigma}\right)}$$

### Vega of Binary Put Option $(\text{Vega}_{BP})$
$$\text{Vega}_{BP}=\frac{\partial V_{BP}}{\partial \sigma}=\frac{\partial}{\partial \sigma}\left(e^{-r\,\left(T-t\right)} N(-d_2)\right)$$
$$=e^{-r\,\left(T-t\right)}n(-d_2)\frac{\partial}{\partial \sigma}\left(-d_2\right)$$
$$=-e^{-r\,\left(T-t\right)}n(d_2)\frac{\partial}{\partial \sigma}\left(d_2\right)$$
Since $n(-x)=n(x)$ and bringing the negative sign from inside to outside the derivative. Now using the work above, from (12), it is clear that
$$\text{Vega}_{BP}=-\text{Vega}_{BC}$$\
$$\Large \implies \boxed{\text{Vega}_{BP}=e^{-r\,\left(T-t\right)}n(d_2)\left(\sqrt{T-t}+\frac{d_2}{\sigma}\right)}$$

## <ins>Rho</ins>
The **rho**,$\,\rho$, of an option is the rate of change of the option value with respect to the interest rate.$$$$
$$\Large \boxed{\Theta=\frac{\partial V}{\partial r}}$$

### Rho of European Call Option $(\rho_{EC})$
$$\rho_{EC}=\frac{\partial V_{EC_{D}}}{\partial r}=\frac{\partial}{\partial r}\left(Se^{-q\,\left(T-t\right)} N(d_1) - Ke^{-r\,\left(T-t\right)} N(d_2)\right)$$
$$=Se^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial r}\left(d_1\right)-K\left( -\left(T-t\right)e^{-r\,\left(T-t\right)} N(d_2)+e^{-r\,\left(T-t\right)} n(d_2)\frac{\partial}{\partial r}\left(d_2\right)\right)$$
$$=Se^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial r}\left(d_1\right)-K\left( -\left(T-t\right)e^{-r\,\left(T-t\right)} N(d_2)+e^{-r\,\left(T-t\right)}\frac{S}{K}n(d_1)e^{\left(r-q\right)\left(T-t\right)} \frac{\partial}{\partial r}\left(d_1-\sigma \sqrt{T-t}\right)\right)$$
Where we have substituted $n(d_2)$ and $d_2$ in terms of $d_1$ derived earlier.\
$$\rho_{EC}=K\left(T-t\right)e^{-r\,\left(T-t\right)} N(d_2)+Se^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial r}\left(d_1\right)-Se^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial r}\left(d_1\right)$$\
$$\Large \implies \boxed{\rho_{EC}=K\left(T-t\right)e^{-r\,\left(T-t\right)} N(d_2)}$$

### Rho of European Put Option $(\rho_{EP})$
$$\rho_{EP}=\frac{\partial V_{EP_{D}}}{\partial r}=\frac{\partial}{\partial r}\left(-Se^{-q\,\left(T-t\right)} N(-d_1) + Ke^{-r\,\left(T-t\right)} N(-d_2)\right)$$
$$=-Se^{-q\,\left(T-t\right)}n(-d_1)\frac{\partial}{\partial r}\left(-d_1\right)+K\left( -\left(T-t\right)e^{-r\,\left(T-t\right)} N(-d_2)+e^{-r\,\left(T-t\right)} n(-d_2)\frac{\partial}{\partial r}\left(-d_2\right)\right)$$
Using the fact $n(-x)=n(x)$ and bringing the negative sign from inside to outside the derivative,$$$$
$$\rho_{EP}=Se^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial r}\left(d_1\right)+K\left( -\left(T-t\right)e^{-r\,\left(T-t\right)} N(-d_2)-e^{-r\,\left(T-t\right)} n(d_2)\frac{\partial}{\partial r}\left(d_2\right)\right)$$
$$=Se^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial r}\left(d_1\right)+K\left( -\left(T-t\right)e^{-r\,\left(T-t\right)} N(-d_2)+e^{-r\,\left(T-t\right)}\frac{S}{K}n(d_1)e^{\left(r-q\right)\left(T-t\right)} \frac{\partial}{\partial r}\left(d_1-\sigma \sqrt{T-t}\right)\right)$$
Where we have substituted $n(d_2)$ and $d_2$ in terms of $d_1$ derived earlier.$$$$
$$\rho_{EP}=-K\left(T-t\right)e^{-r\,\left(T-t\right)} N(-d_2)+Se^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial r}\left(d_1\right)-Se^{-q\,\left(T-t\right)}n(d_1)\frac{\partial}{\partial r}\left(d_1\right)$$\
$$\Large \implies \boxed{\rho_{EP}=-K\left(T-t\right)e^{-r\,\left(T-t\right)} N(-d_2)}$$

### Rho of Binary Call Option $(\rho_{BC})$
$$\rho_{BC}=\frac{\partial V_{BC}}{\partial r}=\frac{\partial}{\partial r}\left(e^{-r\,\left(T-t\right)} N(d_2)\right)$$
$$=-\left(T-t\right)e^{-r\,\left(T-t\right)} N(d_2)+e^{-r\,\left(T-t\right)} n(d_2)\frac{\partial}{\partial r}(d_2)$$
$$=-\left(T-t\right)e^{-r\,\left(T-t\right)} N(d_2)+e^{-r\,\left(T-t\right)} n(d_2)\frac{\partial}{\partial r}\left(\frac{r\left(T-t\right)}{\sigma \sqrt{T-t}}\right)$$
$$=-\left(T-t\right)e^{-r\,\left(T-t\right)} N(d_2)+e^{-r\,\left(T-t\right)} n(d_2)\frac{\sqrt{T-t}}{\sigma}$$\
$$\Large \implies \boxed{\rho_{BC}=-\left(T-t\right)e^{-r\,\left(T-t\right)} N(d_2)+\frac{\sqrt{T-t}}{\sigma}e^{-r\,\left(T-t\right)} n(d_2)}$$

### Rho of Binary Put Option $(\rho_{BP})$
$$\rho_{BP}=\frac{\partial V_{BP}}{\partial r}=\frac{\partial}{\partial r}\left(e^{-r\,\left(T-t\right)} N(-d_2)\right)$$
$$=-\left(T-t\right)e^{-r\,\left(T-t\right)} N(-d_2)+e^{-r\,\left(T-t\right)} n(-d_2)\frac{\partial}{\partial r}(-d_2)$$
$$=-\left(T-t\right)e^{-r\,\left(T-t\right)} N(-d_2)-e^{-r\,\left(T-t\right)} n(d_2)\frac{\partial}{\partial r}(d_2)$$
$$=-\left(T-t\right)e^{-r\,\left(T-t\right)} N(-d_2)-e^{-r\,\left(T-t\right)} n(d_2)\frac{\partial}{\partial r}\left(\frac{r\left(T-t\right)}{\sigma \sqrt{T-t}}\right)$$
$$=-\left(T-t\right)e^{-r\,\left(T-t\right)} N(-d_2)-e^{-r\,\left(T-t\right)} n(d_2)\frac{\sqrt{T-t}}{\sigma}$$\
$$\Large \implies \boxed{\rho_{BC}=-\left(T-t\right)e^{-r\,\left(T-t\right)} N(-d_2)-\frac{\sqrt{T-t}}{\sigma}e^{-r\,\left(T-t\right)} n(d_2)}$$
